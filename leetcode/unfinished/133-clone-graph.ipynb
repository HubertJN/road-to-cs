# 133. Clone Graph

Given a reference of a node in a connected undirected graph, return a deep copy of the graph.

Each node in the graph contains a value `val` and a list of its neighbors.

```python
class Node:
    def __init__(self, val = 0, neighbors = None):
        self.val = val
        self.neighbors = neighbors if neighbors is not None else []
```

## Example 1

**Input:** `adjList = [[2,4],[1,3],[2,4],[1,3]]`  
**Output:** `[[2,4],[1,3],[2,4],[1,3]]`

## Example 2

**Input:** `adjList = [[]]`  
**Output:** `[[]]`

## Example 3

**Input:** `adjList = []`  
**Output:** `[]`

## Constraints

- The number of nodes in the graph is in the range `[0, 100]`.
- `1 <= Node.val <= 100`
- `Node.val` is unique for each node.
- There are no repeated edges and no self-loops in the graph.
- The graph is connected and all nodes can be visited starting from the given node.


````python
from typing import Optional


class Node:
    def __init__(self, val=0, neighbors=None):
        self.val = val
        self.neighbors = neighbors if neighbors is not None else []


class Solution:
    def cloneGraph(self, node: Optional[Node]) -> Optional[Node]:
        pass
````

````python
def build_graph(adj_list):
    if not adj_list:
        return None

    nodes = [Node(i + 1) for i in range(len(adj_list))]
    for i, neighbors in enumerate(adj_list):
        nodes[i].neighbors = [nodes[val - 1] for val in neighbors]
    return nodes[0]


def graph_to_adj_list(node):
    if not node:
        return []

    seen = {}
    stack = [node]
    while stack:
        current = stack.pop()
        if current.val in seen:
            continue
        seen[current.val] = sorted(neighbor.val for neighbor in current.neighbors)
        for neighbor in current.neighbors:
            if neighbor.val not in seen:
                stack.append(neighbor)

    return [seen[i] for i in range(1, len(seen) + 1)]


sol = Solution()

test_cases = [
    [[2, 4], [1, 3], [2, 4], [1, 3]],
    [[]],
    [],
]

for adj_list in test_cases:
    graph = build_graph(adj_list)
    cloned = sol.cloneGraph(graph)
    result = graph_to_adj_list(cloned)
    print(f"input={adj_list}")
    print(f"output={result}")
    print()
````
