# 994. Rotting Oranges

You are given a grid where `0` is an empty cell, `1` is a fresh orange, and `2` is a rotten orange.

Every minute, any fresh orange adjacent 4-directionally to a rotten orange becomes rotten. Return the minimum number of minutes until no fresh oranges remain, or `-1` if this cannot happen.

## Examples

**Input:** `grid = [[2,1,1],[1,1,0],[0,1,1]]`  
**Output:** `4`

**Input:** `grid = [[2,1,1],[0,1,1],[1,0,1]]`  
**Output:** `-1`

**Input:** `grid = [[0,2]]`  
**Output:** `0`

## Constraints

- `m == grid.length`
- `n == grid[i].length`
- `1 <= m, n <= 10`
- `grid[i][j]` is `0`, `1`, or `2`.


```python
from typing import List
from collections import deque

class Solution:
    def orangesRotting(self, grid: List[List[int]]) -> int:
        queue = deque([])
        rows = len(grid)
        columns = len(grid[0])
        fresh_orange = 0
        minutes = -1

        for i in range(rows):
            for j in range(columns):
                orange = grid[i][j]
                if orange == 2:
                    queue.append([i, j])
                elif orange == 1:
                    fresh_orange += 1

        if fresh_orange == 0:
            return 0

        while queue:
            minutes += 1
            for _ in range(len(queue)):
                i, j = queue.popleft()

                for ni, nj in [(i+1, j), (i-1, j), (i, j+1), (i, j-1)]:
                    if (
                        0 <= ni <= rows-1 and 0 <= nj <= columns-1 and
                        grid[ni][nj] == 1
                    ):
                        queue.append([ni, nj])
                        grid[ni][nj] = 2
                        fresh_orange -= 1

        return minutes if fresh_orange == 0 else -1
```

```python
sol = Solution()

test_cases = [
    [[2, 1, 1], [1, 1, 0], [0, 1, 1]],
    [[2, 1, 1], [0, 1, 1], [1, 0, 1]],
    [[0, 2]],
    [[1]],
]

for grid in test_cases:
    result = sol.orangesRotting([row.copy() for row in grid])
    print(f"input={grid}")
    print(f"output={result}")
    print()
```
