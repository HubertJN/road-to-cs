# 15. 3Sum

Given an integer array `nums`, return all unique triplets `[nums[i], nums[j], nums[k]]` where the three values sum to `0` and the indices are distinct.

The answer should not contain duplicate triplets.

## Examples

**Input:** `nums = [-1,0,1,2,-1,-4]`  
**Output:** `[[-1,-1,2],[-1,0,1]]`

**Input:** `nums = [0,1,1]`  
**Output:** `[]`

**Input:** `nums = [0,0,0]`  
**Output:** `[[0,0,0]]`

## Constraints

- `3 <= nums.length <= 3000`
- `-10^5 <= nums[i] <= 10^5`


```python
from typing import List

class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:
        seen = {}
        triplets = []

        for num in nums:
            if num not in seen:
                seen[num] = 1
            else:
                seen[num] += 1

        values = sorted(seen)

        for i in range(len(values)):
            a = values[i]
            for j in range(i, len(values)):
                b = values[j]
                complement = -(a + b)
                if complement not in seen:
                    continue
                if b <= complement:
                    triplet = [a, b, complement]
                    counts = {}
                    valid = True
                    for x in triplet:
                        counts[x] = counts.get(x, 0) + 1
                    for x in triplet:
                        if counts[x] > seen[x]:
                            valid = False
                    if valid:
                        triplets.append(triplet)

        return triplets
```

```python
sol = Solution()

test_cases = [
    [-1, 0, 1, 2, -1, -4],
    [0, 1, 1],
    [0, 0, 0],
    [-2, 0, 1, 1, 2],
]

for nums in test_cases:
    result = sol.threeSum(nums.copy())
    print(f"input={nums}")
    print(f"output={result}")
    print()
```
